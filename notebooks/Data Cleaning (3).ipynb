```python
# Import libraries
import scanpy as sc
import mygene
import pandas as pd
import numpy as np
```

```python
# Read data
adata = sc.read_h5ad("fd28d5c4-34bc-4d28-9afb-61211fe57285.h5ad")
print(adata)
```

```python
# Print observation and variable columns
print(adata.obs.columns)
print(adata.var.columns)
```

```python
# Preview of metadata
adata.obs.head()
```

```python
# Shows the metadata for genes in dataset
print(adata.var.head())
print(adata.var_names[:10])  # Check gene name format
```

```python
# Making the gene IDs readable and usable for data analysis
mg = mygene.MyGeneInfo()
# Query conversion: Ensembl ID → Symbol
query_result = mg.querymany(
    adata.var_names.tolist(),
    scopes='ensembl.gene',
    fields='symbol',
    species='human'
)
```

```python
# Convert query result to DataFrame
query_df = pd.DataFrame(query_result)

# Drop rows without a symbol or duplicate mappings
query_df = query_df.dropna(subset=['symbol']).drop_duplicates(subset=['query'])

# Reorder to match adata.var_names
symbol_map = query_df.set_index('query')['symbol']

# Assign new gene symbols as var_names
adata.var['gene_symbol'] = adata.var_names.map(symbol_map)
adata.var_names = adata.var['gene_symbol']
```

```python
# Restore .var_names to valid strings (Ensembl IDs)
adata.var_names = adata.var.index.astype(str)

# Filter genes that have a valid symbol
valid_mask = ~adata.var['gene_symbol'].isna()
adata = adata[:, valid_mask].copy()
```

```python
# Cell quality control
# Calculate the percentage of mitochondrial genes, which is already available as a fraction.
adata.obs['percent_mt'] = adata.obs['fraction_mitochondrial'] * 100

print("Data shape before filtering:", adata.shape)

# Filter out cells with too few genes (potential empty droplets) or too many (potential doublets).
sc.pp.filter_cells(adata, min_genes=200)
sc.pp.filter_cells(adata, max_genes=6000)

# Filter out cells with a high percentage of mitochondrial DNA (likely stressed or dying cells).
adata = adata[adata.obs.percent_mt < 10, :]

print("Data shape after filtering:", adata.shape)
```

```python
# Gene quality control (no genes were filtered)
print(f"Shape before gene filtering: {adata.shape}")
# Make variable names (gene symbols) unique to resolve warnings.
adata.var_names_make_unique()

# Filter out genes that are expressed in fewer than 3 cells.
sc.pp.filter_genes(adata, min_cells=3)
print(f"Shape after gene filtering: {adata.shape}")
```

```python
# Create the C matrix (cell_type x genes)
C_matrix = (
    pd.DataFrame(adata.X.toarray(), index=adata.obs["cell_type"], columns=adata.var_names)
    .groupby(level=0)
    .mean()
)
```

```python
# Build the p matrix

# Count number of cells per (sample, cell type)
cell_counts = (
    adata.obs
    .groupby(['sample_id', 'cell_type'])
    .size()
    .unstack(fill_value=0)  # Rows = samples, Columns = cell types
)

# Normalize to get proportions (each row sums to 1)
P_matrix = cell_counts.div(cell_counts.sum(axis=1), axis=0)

# Preview
print(P_matrix.head())

```

```python
# Build the T matrix 

# Step 1: Convert X to dense matrix
X_dense = adata.X.toarray() if not isinstance(adata.X, np.ndarray) else adata.X

# Step 2: Attach sample_id as a new column (not index)
X_df = pd.DataFrame(X_dense, columns=adata.var_names)
X_df['sample_id'] = adata.obs['sample_id'].values

# Step 3: Group by sample_id and sum across cells to simulate pseudo-bulk
T_matrix = X_df.groupby('sample_id').sum()

# Step 4: Preview
print(T_matrix.head())
```

```python

# CONFIGURABLE
n_samples = 100
cells_per_sample = 1000  # or vary this across samples

# Setup
X_dense = adata.X.toarray() if not isinstance(adata.X, np.ndarray) else adata.X
cell_types = adata.obs["cell_type"].values
sample_indices = np.arange(X_dense.shape[0])
gene_names = adata.var_names
unique_cell_types = np.unique(cell_types)

# Store outputs
T_simulated = []
P_simulated = []

for _ in range(n_samples):
    # Randomly sample cells
    selected_idx = np.random.choice(sample_indices, size=cells_per_sample, replace=True)
    expr = X_dense[selected_idx, :]  # gene expression
    types = cell_types[selected_idx]  # corresponding cell types
    
    # Build T: sum expression
    T_simulated.append(expr.sum(axis=0))
    
    # Build P: proportions
    counts = pd.Series(types).value_counts(normalize=True).reindex(unique_cell_types, fill_value=0)
    P_simulated.append(counts.values)

# Convert to DataFrames
T_matrix = pd.DataFrame(T_simulated, columns=gene_names)
P_matrix = pd.DataFrame(P_simulated, columns=unique_cell_types)

# Preview
print("T_matrix shape:", T_matrix.shape)
print("P_matrix shape:", P_matrix.shape)
T_matrix.head(), P_matrix.head()
```

```python
# Data normalization

# Store raw data
adata.raw = adata.copy()

# --- Before Normalization ---
print("--- Before Normalization (from adata.raw) ---")
# To access the raw data, use the .raw attribute
print(f"Sum of counts per cell: {adata.raw.X.sum(axis=1)[:5]}")
# Display a slice of the raw integer counts
print(f"Data looks like: {adata.raw.X[:2, :10].toarray().flatten()}")


# Normalize each cell to have the same total UMI count, so expression levels are comparable.
sc.pp.normalize_total(adata, target_sum=1e4)

# Log-transform the data to stabilize variance and make patterns more visible.
sc.pp.log1p(adata)

# --- After Normalization ---
print("\n--- After Normalization (from adata) ---")
# Now, check the modified adata.X
print(f"Sum of counts per cell: {adata.X.sum(axis=1)[:5]}")
# Display a slice of the transformed float data
print(f"Data looks like: {adata.X[:2, :10].toarray().flatten().round(2)}")

print("\nFinal cleaned and normalized data object:")
print(adata)
```
