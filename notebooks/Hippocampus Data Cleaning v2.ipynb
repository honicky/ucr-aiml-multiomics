```python
# Making the gene IDs readable and usable for data analysis
mg = mygene.MyGeneInfo()
# Query conversion: Ensembl ID → Symbol
query_result = mg.querymany(
    adata.var_names.tolist(),
    scopes='ensembl.gene',
    fields='symbol',
    species='human'
)
```

```python
# Convert query result to DataFrame
query_df = pd.DataFrame(query_result)

# Drop rows without a symbol or duplicate mappings
query_df = query_df.dropna(subset=['symbol']).drop_duplicates(subset=['query'])

# Reorder to match adata.var_names
symbol_map = query_df.set_index('query')['symbol']

# Assign new gene symbols as var_names
adata.var['gene_symbol'] = adata.var_names.map(symbol_map)
adata.var_names = adata.var['gene_symbol']
```

```python
# Restore .var_names to valid strings (Ensembl IDs)
adata.var_names = adata.var.index.astype(str)

# Filter genes that have a valid symbol
valid_mask = ~adata.var['gene_symbol'].isna()
adata = adata[:, valid_mask].copy()
```

```python
# Cell quality control
# Calculate the percentage of mitochondrial genes, which is already available as a fraction.
adata.obs['percent_mt'] = adata.obs['fraction_mitochondrial'] * 100

print("Data shape before filtering:", adata.shape)

# Filter out cells with too few genes (potential empty droplets) or too many (potential doublets).
sc.pp.filter_cells(adata, min_genes=200)
sc.pp.filter_cells(adata, max_genes=6000)

# Filter out cells with a high percentage of mitochondrial DNA (likely stressed or dying cells).
adata = adata[adata.obs.percent_mt < 10, :]

print("Data shape after filtering:", adata.shape)
```

```python
# Gene quality control (no genes were filtered)
print(f"Shape before gene filtering: {adata.shape}")
# Make variable names (gene symbols) unique to resolve warnings.
adata.var_names_make_unique()

# Filter out genes that are expressed in fewer than 3 cells.
sc.pp.filter_genes(adata, min_cells=3)
print(f"Shape after gene filtering: {adata.shape}")
```

```python
# Create the C matrix (cell_type x genes)
C_matrix = (
    pd.DataFrame(adata.X.toarray(), index=adata.obs["cell_type"], columns=adata.var_names)
    .groupby(level=0)
    .mean()
)
```

```python
# Create a cell count matrix based on sample ID before making the P matrix

# Count number of cells per (sample, cell type)
cell_counts = (
    adata.obs
    .groupby(['sample_id', 'cell_type'])
    .size()
    .unstack(fill_value=0)  # Rows = samples, Columns = cell types
)

# print a preview before it becomes a proportion to see the cell count matrix
print(cell_counts.head())
```

```python
#Building the P matrix from the cell count matrix

# Normalize to get proportions (each row sums to 1)
P_matrix = cell_counts.div(cell_counts.sum(axis=1), axis=0)

# Preview the P matrix
print(P_matrix.head())
```

```python
# Build the T matrix with a sparse matrix

# Step 1: Create a sparse DataFrame directly from the sparse X matrix
# This avoids creating a dense matrix in memory.
X_df_sparse = pd.DataFrame.sparse.from_spmatrix(
    adata.X,
    index=adata.obs_names,
    columns=adata.var_names
)

# Step 2: Add the sample_id for grouping
X_df_sparse['sample_id'] = adata.obs['sample_id'].values

# Step 3: Group by sample_id and sum the counts
# Pandas performs this operation efficiently on the sparse DataFrame.
T_matrix = X_df_sparse.groupby('sample_id').sum()

# Step 4: Preview the result
print(T_matrix.head())
```

```python
from sklearn.model_selection import train_test_split

# --- 1. Split donor_ids into training and testing sets ---
unique_donors = adata.obs['donor_id'].unique()
train_donors, test_donors = train_test_split(unique_donors, test_size=0.2, random_state=42)

# --- 2. Get cell indices corresponding to the train/test donors ---
train_indices = adata.obs.index[adata.obs['donor_id'].isin(train_donors)]
test_indices = adata.obs.index[adata.obs['donor_id'].isin(test_donors)]

# Convert to numerical indices that can be used to slice adata.X
train_indices_loc = [adata.obs_names.get_loc(i) for i in train_indices]
test_indices_loc = [adata.obs_names.get_loc(i) for i in test_indices]

# --- 3. Use the simulation function from before ---
def simulate_pseudobulk_with_error(cell_indices, n_samples, cells_per_sample=1000, noise_level=5.0):
    """Generates T and P matrices, adding random noise to T."""
    T_simulated = []
    P_simulated = []

    X_sparse = adata.X
    cell_types = adata.obs["cell_type"].values
    unique_cell_types = np.unique(cell_types)
    gene_names = adata.var_names

    for _ in range(n_samples):
        selected_idx = np.random.choice(cell_indices, size=cells_per_sample, replace=True)
        expr = X_sparse[selected_idx, :]
        types = cell_types[selected_idx]

        T_simulated.append(expr.sum(axis=0))

        counts = pd.Series(types).value_counts(normalize=True).reindex(unique_cell_types, fill_value=0)
        P_simulated.append(counts.values)

    T_matrix = pd.DataFrame(np.vstack(T_simulated), columns=gene_names)
    P_matrix = pd.DataFrame(P_simulated, columns=unique_cell_types)

    noise = np.random.normal(0, noise_level, size=T_matrix.shape)
    T_matrix_noisy = T_matrix + noise
    T_matrix_noisy[T_matrix_noisy < 0] = 0

    return T_matrix_noisy, P_matrix

# --- 4. Generate the independent training and testing sets ---
T_train, P_train = simulate_pseudobulk_with_error(train_indices_loc, n_samples=80)
T_test, P_test = simulate_pseudobulk_with_error(test_indices_loc, n_samples=20)

print("Training set created from donors:", train_donors)
print("Testing set created from donors:", test_donors)
print("-" * 20)
print("T_train shape:", T_train.shape)
print("T_test shape:", T_test.shape)

# Print the first 5 rows of the training matrices
print("--- Training T Matrix (T_train) ---")
print(T_train.head())

print("\n--- Training P Matrix (P_train) ---")
print(P_train.head())

# Print the first 5 rows of the testing matrices
print("\n--- Testing T Matrix (T_test) ---")
print(T_test.head())

print("\n--- Testing P Matrix (P_test) ---")
print(P_test.head())
```
